## 1. Keras

Keras is a high-level neural network API written in Python. It is designed for fast experimentation with deep learning models and is user-friendly, modular and extensible.

**Distinct features and functionality**
- Easy to learn, with a simple and consistent API.
- Reduces the amount of code required to build common neural-network models.
- Supports quick prototyping and experimentation.
- Runs on CPU and GPU.
- Supports many neural-network architectures.
- Provides Sequential and Functional model-building approaches.
- Works closely with TensorFlow in modern applications.

**Steps to build a model using Keras**
1. Define the network and its layers.
2. Compile the network by specifying the loss function, optimizer and metrics.
3. Fit the model to training data.
4. Evaluate the model.
5. Make predictions using `model.predict()`.

**Common functionality:** defining layers, compiling models, training, evaluating and predicting.

````python
# Check whether TensorFlow and its Keras API are available
try:
    import tensorflow as tf
    print("TensorFlow version:", tf.__version__)
    print("Keras is available through tf.keras")
except ImportError:
    print("TensorFlow/Keras is not installed in this Python environment.")
````

## 2. PyTorch

PyTorch is an optimized deep-learning tensor library based on Python and Torch. It supports CPU and GPU computation and uses dynamic computation graphs.

**Distinct features and functionality**
- Tensor computation similar to NumPy, with GPU support.
- Automatic differentiation for calculating gradients.
- Dynamic computation graphs, making debugging and experimentation easier.
- Supports mathematical operations such as addition, subtraction, multiplication and division.
- Supports matrix initialization and matrix operations.

**Common PyTorch modules**
- `torch.nn`: classes and layers used to build neural-network models.
- `torch.autograd`: automatic differentiation and gradient calculation.
- `torch.optim`: optimization algorithms used to update model parameters.

````python
# Basic PyTorch tensor demonstration
try:
    import torch
    print("PyTorch version:", torch.__version__)

    a = torch.tensor([1.0, 2.0, 3.0])
    b = torch.tensor([4.0, 5.0, 6.0])

    print("Tensor A:", a)
    print("Tensor B:", b)
    print("Addition:", a + b)
    print("Multiplication:", a * b)
except ImportError:
    print("PyTorch is not installed in this Python environment.")
````

## 3. TensorFlow

TensorFlow is an open-source library developed by Google for deep learning and other machine-learning applications. It accepts data in the form of multidimensional arrays called tensors.

**Distinct features and functionality**
- Supports tensor-based numerical computation.
- Supports CPU and GPU execution.
- Uses data-flow graphs to represent operations and the flow of data.
- Supports large-scale and distributed computation.
- Can be used to build, train and evaluate deep-learning models.

**TensorFlow architecture – main steps**
1. Data preprocessing: prepare and structure the input data.
2. Model building: define the model architecture.
3. Training and estimation: train the model and test it using unknown data.

**Important components**
- **Tensor:** A multidimensional data array used in computations.
- **Graph:** Represents operations as nodes and the connections between them as edges.

````python
# Basic TensorFlow tensor demonstration
try:
    import tensorflow as tf
    print("TensorFlow version:", tf.__version__)

    x = tf.constant([1, 2, 3])
    y = tf.constant([4, 5, 6])

    print("Tensor X:", x.numpy())
    print("Tensor Y:", y.numpy())
    print("Addition:", (x + y).numpy())
    print("Multiplication:", (x * y).numpy())
except ImportError:
    print("TensorFlow is not installed in this Python environment.")
````

## 4. Theano

Theano is a Python library for efficient numerical computation. It can evaluate mathematical operations on multidimensional arrays and supports CPU and GPU computation.

**Distinct features and functionality**
- Efficient numerical computation on multidimensional arrays.
- Optimizes expressions for better execution speed.
- Supports GPU execution.
- Performs symbolic differentiation to calculate gradients automatically.
- Supports computations used in convolutional and recurrent neural networks.

**Installation command from the lab manual**
```bash
pip install theano
```

Theano is a legacy deep-learning library. On modern Python environments, installation or execution may not be supported. The study and features above are included as described in the lab manual.

````python
# Check whether Theano is available
try:
    import theano
    import theano.tensor as T
    print("Theano version:", theano.__version__)

    a = T.dscalar('a')
    b = T.dscalar('b')
    add_expr = a + b
    add_function = theano.function([a, b], add_expr)
    print("Theano addition (10 + 20):", add_function(10, 20))
except Exception as e:
    print("Theano is not available or could not be imported in this environment.")
    print("Reason:", str(e)[:250])
````
